```python
df_price = pd.read_csv('data/price_data.csv')
df_price
```

```python
# incorporate this into mother - for loop with the name of the "primary_commodity" column - "Other (poly)-metallic" = "Other" in price 
```

```python
mean = df_price.mean()
std = df_price.std()
variation = std*100 / mean
# print(mean, std, variation)

variation = variation[1:]
variation
```

```python
# encoding average price variation into average_data (created after much of this code)
average_data = pd.read_csv('data/average_data.csv')

ohe_cols = [col for col in average_data.columns if col.startswith("primary_commodity_")]

def get_variation(row):
    for col in ohe_cols:
        if row[col] == 1:
            commodity = col.replace("primary_commodity_", "")
            return variation.get(commodity, np.nan)
    return np.nan
```

```python
average_data["commodity_var"] = average_data.apply(get_variation, axis=1)
```
